```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv("/content/titanic.csv")
df
df.head()
```

```python
print(df.shape)      #To Show The No Of Rows & Columns
```

```python
df.info()           #It Shows The Data Type,Column Names
```

```python
df.describe()               #Statistical Summary Of Numerical Columns,Including Mean,Median,Mode
```

```python
df.isnull().sum()
```

```python
df["Embarked"]=df["Embarked"].fillna(        #Replace Missing Value  |  #Find Freq Values
    df["Embarked"].mode()[0]
)
```

```python
print(df.isnull().sum())
print("Before:",df.shape[0])
df=df.drop_duplicates()
```

```python
print("After:",df.shape[0])       #Rows Count (0)
df=df.rename(columns={
    "Passengerid": "Passenger ID",
    "2urvived": "Survived"
})
```

```python
df["Sex"]=df["Sex"].map({
    0: "Male",
    1: "Female"
})
df
```

```python
sns.histplot(data=df,x="Age",bins=20)
plt.title("Age Distribution Of Titanic Passenger")
plt.xlabel("Age")
plt.ylabel("Number Of Passeneger")
plt.show()
sns.countplot(data=df,x="Sex")
plt.title("Titanic Survival By Sex")
plt.xlabel("Sex")
plt.ylabel("Number Of Passengers")
plt.show()
sns.countplot(data=df,x="Pclass")
plt.title("Titanic Survival By Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number Of Passengers")
plt.show()
```
