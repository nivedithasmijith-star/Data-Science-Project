#              **MATPLOTLIB**

```python

import matplotlib.pyplot as plt
x=[2,4,6,8]
y=[1,3,4,7]
plt.plot(x,y)
plt.show()           #Display the Graph
```

```python
import matplotlib.pyplot as plt
months=["Jan","Feb","Mar","Apr","May"]
sales=[20,30,25,40,50]

plt.plot(months,sales)
plt.show()
```

```python
import matplotlib.pyplot as plt

months=["Jan","Feb","Mar","Apr","May"]
sales=[20,30,25,40,50]
profit=[5,8,6,12,15]

plt.plot(months,sales,label="Sales")              #Label
plt.plot(months,profit,label="Profit")
plt.legend()                                      #Display the Label

plt.show()
```

```python
import matplotlib.pyplot as plt

months=["Jan","Feb","Mar","Apr","May"]
sales=[20,30,25,40,50]
profit=[5,8,6,12,15]

plt.xlabel("Month",color="purple")
plt.ylabel("Sales",color="purple")
plt.title("Monthly Sales",color="red")
plt.grid(False)
plt.plot(
    months,
    sales,
    marker="s",
    color="green",
    linestyle="--"
)
plt.show()
```

**BAR CHART**

```python
import seaborn as sns
import matplotlib.pyplot as plt

months=["Jan","Feb","Mar","Apr","May"]
sales=[20,30,25,40,50]
sns.barplot(sales,color="purple")
plt.title("Monthly Sales",color="green")
plt.show()
```

```python
import matplotlib.pyplot as plt

months=["Jan","Feb","Mar","Apr","May"]
sales=[20,30,25,40,50]
profit=[5,8,6,12,15]
plt.bar(months,sales,label="Sales")          #Seperation for sales and profit
plt.bar(months,profit,label="Profit")

plt.legend()

```

**SCATTER**

```python
import matplotlib.pyplot as plt

study_hours=[1,2,3,4,5,6]
marks=[40,45,55,65,75,85]

plt.scatter(study_hours,marks,color="purple")


plt.title("study_hours vs marks",color="green")
plt.xlabel("study_hours",color="blue")
plt.ylabel("marks",color="red")
plt.show()
```

**HISTOGRAM**

```python
import matplotlib.pyplot as plt

marks=[45,50,52,55,60,62,65,68,70,72,75,78,80,82,85,90]

plt.hist(marks,color="purple",edgecolor="yellow")


plt.title("marks distribution",color="purple")
plt.xlabel("marks",color="orange")
plt.ylabel("frequency",color="orange")


plt.show()
```

**PIE CHART**

```python
import matplotlib.pyplot as plt
categories=["Food", "Travel", "Shopping","Others"]
amount=[4000,2000,3000,1000]

plt.pie(amount,labels=categories)
plt.title("Monthly Expenses")

plt.show()
```

**FIG**

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(2,1)                              # fig is the entire plotting area
                                                         #subplot is to diplay 2 graphs in 1 area.
ax[0].plot([1,2,3],[10,20,30],color="green")              # ax is individual graphs
ax[1].bar(["A","B","C"],[20,15,30],color="purple")

plt.show()
```
