# ***SEABORN***

```python
import seaborn as sns
import matplotlib.pyplot as plt
data={
    "Product":["Laptop","Phone","Tablet","Laptop","Phone","Tablet"],
    "Sales":[50,80,40,70,90,60],
    "Price":[50000,30000,20000,55000,32000,22000],
    "Month":["Jan","Jan","Feb","Feb","Mar","Mar"],
    }
#BARPLOT
sns.barplot(x="Product",y="Sales",data=data)          #Compare sales of different products
plt.show()

#COUNTPLOT
sns.countplot(x="Product",data=data)                  #Count how many times a product appears
plt.show()

#BOXPLOT
sns.boxplot(y=data["Sales"])                         #Show the spread of sales and possible outliers
plt.show()

#SCATTERPLOT
sns.scatterplot(x="Price",y="Sales",data=data)       #Rltion btw sales & price
plt.show()

#HISTPLOT
sns.histplot(x="Product",y="Sales",data=data)
plt.show()

#LINEPLOT
sns.lineplot(x="Month",y="Sales",data=data)           #Show the trend by month
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
data=np.random.randn(1000)
plt.figure(figsize=(8,6))
sns.histplot(data,bins=50,kde=True,color="cyan",edgecolor="red")         #kernel Density Estimation(curve)
plt.xlabel("values")
plt.ylabel("frequency")
plt.title("Distribution Of Random Data Using Seaborn")
plt.grid(axis="y",alpha=0.75)
plt.show()
                                                                          #markers=., * , o , s , p , ^     &      linespace = '- , -. , : , --- , --
```

```python
products=["Laptop","Phone","Tablet"]
Sales=[50,80,40]
plt.pie(Sales, labels=products,autopct="%4.2f%%")
plt.title("Sales Distribution")
plt.show()
```
