```python
import pandas as pd
data=[20,20,30,30,50,60,60,80]
ar=pd.Series(data)
print(ar)
print(ar.unique())         #TO REMOVE DUPLICATES
print(ar.nunique())        #NO OF UNIQUE ITEMS
print(ar.value_counts())   #NO OF TIMES EACH ITEM IS REPEATED
```
